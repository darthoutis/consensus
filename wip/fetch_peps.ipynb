```python
import requests
from datetime import datetime
import tiktoken
import json

with open("config.json") as f:
    config = json.load(f)

gh_token = config["github_token"]

headers = {"Authorization": f"Bearer {gh_token}"}
tokenizer = tiktoken.get_encoding("cl100k_base")

url = "https://peps.python.org/api/peps.json"
peps_data = requests.get(url).json()

# peps_data = requests.get(url,headers=headers).json()

TARGET_DATE = datetime.fromisoformat("2018-08-01")

matching_peps = []

for pep_num, pep_info in peps_data.items():
    # Parse creation date from metadata
    created_str = pep_info.get("created")  # e.g., "01-Aug-2018" or "2018-08-01"
    if created_str:
        try:
            # Handle various date formats stored in PEP headers
            created_dt = datetime.strptime(created_str, "%d-%b-%Y")
            if created_dt >= TARGET_DATE:
                matching_peps.append((f"pep-{pep_num.zfill(4)}", created_str))
        except ValueError:
            pass

print(f"Found {len(matching_peps)} matching PEPs in 1 request.")
for pep, date in matching_peps:
    print(f"{pep} (Created: {date})")
```
